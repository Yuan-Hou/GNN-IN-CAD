# Understanding the Contribution of B-Rep Graph Topology to UV-Net’s Strong Performance

## Introduction

### Synopsis

UV-Net’s original ablation on MFCAD reported an IoU of 85.34% using face geometry alone, compared with 99.87% for the full geometry-and-topology model [1]. This project investigates that improvement through a controlled model ladder for face-level machining-feature classification on MFCAD [2]. The ladder separates the contributions of shallow topology information, learned message passing, joint optimization of geometry and graph models, and edge conditioning.

Starting with a face CNN and an independent face classifier, we augment the geometry embeddings with shallow topology features without learned message passing. We then introduce a GCN, first with frozen geometry embeddings and then with joint end-to-end training, before adding edge geometry and edge-conditioned message passing. Advanced architectures are optional extensions.

Aggregate performance, per-class metrics, confusion patterns, and learned-embedding visualizations will help identify where each stage adds useful information. We relate these findings to geometric and graph properties, examining when geometry alone is sufficient and when graph context becomes informative. The aim is an evidence-based decomposition of UV-Net’s strong performance on MFCAD, rather than a definitive mechanistic explanation.

### References

[1] Jayaraman, P. K., et al. (2021). *UV-Net: Learning From Boundary Representations*. CVPR, 11703–11712.

[2] Cao, W., et al. (2020). *Graph Representation of 3D CAD Models for Machining Feature Recognition With Deep Learning*. ASME IDETC-CIE.
