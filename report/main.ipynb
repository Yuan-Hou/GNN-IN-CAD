# Understanding the Contribution of B-Rep Graph Topology to UV-Net’s Strong Performance

## Introduction

### Synopsis

UV-Net’s original MFCAD ablation reported an IoU of 85.34% using face geometry alone, compared with 99.87% for its full geometry-and-topology model [1]. We investigate what B-rep graph information contributes to this gap through face-level machining-feature classification on the synthetic MFCAD dataset [2]. Our aim is to interpret the gains from adding graph information using a simple, controlled model ladder, rather than reproduce UV-Net exactly or maximize its score.

Using the UV-Net authors’ prepared data, we examine both sampled face geometry and graph structure. EDA identifies geometric cues such as span and normal alignment, alongside degree patterns and neighbourhood reachability that motivate the graph comparisons. Visual inspection also reveals a mismatch between supplied class names and observed geometry; we establish a reviewed name mapping before interpreting class-level results.

Starting from an independent face CNN classifier, the ladder adds shallow topology descriptors, learned message passing with frozen and jointly trained face embeddings, and finally boundary-curve geometry with edge-conditioned messages. Validation mean IoU over the 15 machining-feature classes, excluding stock, rises from **78.05% for geometry alone** to **91.82% with graphlets**, **98.86% with joint GCN**, and **99.92% with updated edge conditioning**.

Embedding views, property distributions, confusion tables and whole-solid examples connect these improvements back to the EDA. The results suggest that local connectivity resolves many shape confusions, neighbouring-face geometry distinguishes features with similar face degrees, and boundary geometry helps with difficult intersecting and oblique passage cases. These observations offer an evidence-based account of the roles of node, neighbourhood and edge information, while remaining interpretations from single-run experiments on a constrained synthetic dataset.

### References

[1] Jayaraman, P. K., et al. (2021). *UV-Net: Learning From Boundary Representations*. CVPR, 11703–11712.

[2] Cao, W., et al. (2020). *Graph Representation of 3D CAD Models for Machining Feature Recognition With Deep Learning*. ASME IDETC-CIE.
