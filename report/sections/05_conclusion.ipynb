# Conclusion

This project contributes a connected investigation of what graph information adds to face-level machining-feature classification on MFCAD:

1. **A controlled model ladder.** We separate shallow topology descriptors, learned message passing, joint geometry training and edge conditioning, showing that progressively adding graph information produces the expected improvement in IoU.
2. **Transition-specific correction analysis.** Comparing full confusion matrices one by one makes it difficult to see what changed. We instead isolate faces that were **wrong in the previous stage and correct in the next**, then examine their properties and class-to-class confusions to identify what each step resolves.
3. **Information-dense visual comparisons.** Embedding panels make class mixing and separation easier to spot across models, while heatmap colouring highlights dominant confusion routes without emphasizing low-support rows. Whole-solid examples connect these patterns to the underlying geometry.

The results suggest two broad sources of improvement: degree and local connectivity resolve many shape confusions, while learned message passing adds information about neighbouring faces that counts alone cannot distinguish. Edge conditioning addresses much of the remaining passage confusion. Together, the ladder raises non-stock validation mean IoU from **78.05% to 99.92%**; the analysis explains where those gains occur in terms of node, neighbourhood and boundary information.

## EDA as the groundwork for analysis

EDA established our approach of examining **geometry-related properties first, then graph-related properties**, laying the groundwork for the later analysis. Face spans and normal alignment identified cues available from individual faces; degree patterns and neighbourhood reachability exposed structural information available through the graph. Returning to these same properties in the corrected subsets connected the model comparisons to the data, rather than interpreting scores in isolation. The resulting patterns vary across stages, with no uniform benefit for smaller, oblique or non-degree-four faces.

EDA also informed model building: neighbourhood reachability showed that a face reaches a median 73.9% of other faces within two hops and all other faces within three, supporting our choice of two hops for the main graph models.

Finally, unexpected geometric statistics prompted inspection of two examples per class, confirming a label-name mismatch that we corrected before it led to misleading interpretations.

## Scope and limitations

- **Dataset scope:** MFCAD is synthetic, with constrained machining operations and strong geometric regularities. Near-saturated performance does not establish equivalent behaviour on more varied industrial CAD models.
- **Experimental scope:** Comparisons use single runs and final-epoch validation results.
- **Correction analysis:** The selected subsets assume newly introduced errors are a minority and show corrections rather than net gains.
- **Performance saturation:** Near-perfect performance leaves little room to distinguish further improvements, limiting our investigation of more advanced models such as Graph Transformers.
