# Conclusion

Our controlled model ladder shows that much of the gap between geometry-only classification and near-perfect MFCAD performance can be recovered by progressively adding B-rep graph information. Validation mean IoU over the 15 machining-feature classes rises from **78.05% with face geometry alone**, to **91.82% with shallow graphlet features**, **98.86% with jointly trained GCN message passing**, and **99.92% with updated edge conditioning**. The simpler tutorial-based setup makes these changes easier to interpret without requiring a full reproduction of UV-Net.

## EDA as the foundation of the investigation

EDA established the geometry-versus-graph framework used throughout the project. Face spans and normal alignment revealed simple geometric cues available to the face CNN, while degree patterns and neighbourhood reachability identified information that individual faces cannot supply. This gave the model ladder concrete questions to test: how much local topology adds, what learned exchanges between neighbouring faces resolve, and when boundary-curve geometry becomes useful.

The graph exploration also informed the range of receptive fields we considered. A face reaches a median 73.9% of the other faces within two hops and all other faces within three. This made two-hop models a reasonable compact starting point and three hops a meaningful comparison, rather than treating depth as an unrestricted tuning exercise. The final two-hop choice was supported by the model comparisons; the common optimizer and training settings followed the tutorial-based experimental setup.

Crucially, inspecting geometry exposed a mismatch between the supplied class names and the observed machining features. Eyeballing two examples per class allowed us to establish a one-to-one name mapping before interpreting the class-level results. Without this check, apparent observations such as axis-aligned “chamfers” would have led to misleading conclusions about both the dataset and the models. The reviewed names then carried consistently through EDA, model reporting, embeddings and confusion analysis.

## What the ladder helps explain

The subsequent analysis returned to the same geometric and graph cues rather than stopping at aggregate scores. Shallow topology corrected many reciprocal shape confusions, consistent with local degree patterns distinguishing pockets and passages. GCN message passing resolved more asymmetric feature-type confusions, including the through-step → through-slot → passage and blind-step → blind-slot → pocket chains, where similar rectangular faces limit what degree counting alone can distinguish. Edge conditioning corrected the remaining difficult passage cases, plausibly by supplying boundary directions and geometry when intersections and oblique faces disrupt the usual cues.

The interpretation is also explicitly graph-based. Degree and graphlet features describe local connectivity without learning how neighbouring geometry should contribute. They can distinguish some feature structures, but a rectangular face with four boundaries can occur in several different machining features. Learned message passing adds the geometry of neighbouring faces and their arrangement within the receptive field, helping distinguish faces that share the same local degree. Edge conditioning further makes those exchanges depend on the sampled boundary connecting each pair of faces, rather than treating adjacency alone as the relationship. The step–slot–passage examples therefore illustrate differences in the information available on the graph, as well as differences between machining operations.

This perspective also guided how we read the property plots. The larger two- and three-hop neighbourhoods among edge-stage corrections suggest that boundary geometry helps interpret more complex connectivity. Conversely, the rise and fall of geometric spans across corrected subsets need not mean that a graph variant is intrinsically better at a particular face shape: graph structures and face geometry are correlated. Our conclusions connect observed errors to node information, neighbourhood structure and edge information, while treating the proposed mechanisms as interpretations rather than causal proof.

Embeddings suggested where classes remained entangled; difficult faces made those ambiguities concrete; property distributions related corrections to geometry and graph context; and confusion tables identified the class pairs involved. Together, these views support an interpretation in which graph information progressively disambiguates locally similar faces. Shifts in the geometric distributions of corrected faces reflect which structural patterns benefit, rather than proving that a graph variant directly solves a particular geometric challenge.

## Scope and limitations

- **Dataset scope:** MFCAD is synthetic, with constrained machining operations and strong geometric regularities. Near-saturated performance does not establish equivalent behaviour on more varied industrial CAD models.
- **Experimental scope:** Comparisons use single runs and final-epoch validation results.
- **Correction analysis:** The selected subsets assume newly introduced errors are a minority and show corrections rather than net gains.
- **Architectural attribution:** Edge conditioning also changes the graph operator, so its gain cannot be attributed exclusively to curve features.

Within that scope, the contribution is a connected investigation: checking the data and its labels, using EDA to frame a compact model ladder, and returning to those same properties to interpret its improvements. The result is both a performance comparison and an interpretation of what information the graph contributes: local connectivity, neighbouring-face geometry, and the geometry of the boundaries through which messages pass.
